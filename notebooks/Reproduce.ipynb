# VITAL Q reproduction
Run the published analysis, a full fresh campaign, or both. A fresh full run can take hours. Results are written to a new `runs/` directory; published inputs remain unchanged. Install the repository requirements in this kernel environment first.

```python
from pathlib import Path
import sys, subprocess, time, uuid
REPO = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "scripts/reproduce.py").exists())
OPTION = 1  # 1: published numerical replay; 2: full fresh campaign; 3: both plus proof verification
MODE = {1: "replay", 2: "fresh", 3: "all"}[OPTION]
OUTPUT = REPO / "runs" / ("notebook-" + MODE + "-" + uuid.uuid4().hex[:8])
print("Mode:", MODE, "\nOutput:", OUTPUT)
```

```python
subprocess.run([sys.executable, str(REPO / "scripts/reproduce.py"), MODE, "--output", str(OUTPUT)], check=True)
print("Completed:", OUTPUT)
```

```python
from IPython.display import display, Image
for figure in sorted(OUTPUT.rglob("prediction_errors.png")):
    print(figure.relative_to(OUTPUT))
    display(Image(filename=str(figure)))
```
